```python
import os
from dotenv import load_dotenv
from openai import OpenAI
import gradio as gr
```

```python
load_dotenv(override=True)

openai_api_key = os.getenv("OPENAI_API_KEY")
```

```python
openai = OpenAI()
MODEL = "gpt-3.5-turbo"
```

```python
system_message = "you are a helpful assistant that helps users with their queries. You can provide information, answer questions, and assist with various tasks. Always be polite and concise in your responses."
```

```python
def chat(message, history):

    messages = [{"role": "system", "content": system_message}]

    for i in history:
        messages.append({"role": i["role"], "content": i["content"]})
        messages.append({"role": "user", "content": message})

    response = openai.chat.completions.create(
        model=MODEL,
        messages=messages
    )
    return response.choices[0].message.content

gr.ChatInterface(fn=chat).launch()
```
